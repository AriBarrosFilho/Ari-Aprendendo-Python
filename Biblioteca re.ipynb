```python
import re

# ============================================
# GUIA COMPLETO DE REGEX EM PYTHON
# ============================================

print("=" * 60)
print("1. METACARACTERES BÁSICOS")
print("=" * 60)

# . (ponto) - qualquer caractere exceto nova linha
texto = "gato, gito, guto, g@to"
print(f"\nTexto: {texto}")
print(f"Padrão r'g.to': {re.findall(r'g.to', texto)}")

# ^ - início da string
texto = "Python é ótimo. Python rocks!"
print(f"\nTexto: {texto}")
print(f"Padrão r'^Python': {re.findall(r'^Python', texto)}")

# $ - fim da string
print(f"Padrão r'rocks!$': {re.findall(r'rocks!$', texto)}")

# * - 0 ou mais repetições
texto = "a aa aaa aaaa"
print(f"\nTexto: {texto}")
print(f"Padrão r'a*': {re.findall(r'a+', texto)}")

# + - 1 ou mais repetições
texto = "goool gol goooool"
print(f"\nTexto: {texto}")
print(f"Padrão r'go+l': {re.findall(r'go+l', texto)}")

# ? - 0 ou 1 repetição (opcional)
texto = "color colour"
print(f"\nTexto: {texto}")
print(f"Padrão r'colou?r': {re.findall(r'colou?r', texto)}")

print("\n" + "=" * 60)
print("2. QUANTIFICADORES")
print("=" * 60)

# {n} - exatamente n repetições
texto = "O número é 123 e 12345 e 12"
print(f"\nTexto: {texto}")
print(f"Padrão r'\\d{{3}}' (exatamente 3 dígitos): {re.findall(r'\d{3}', texto)}")

# {n,m} - de n a m repetições
print(f"Padrão r'\\d{{2,4}}' (de 2 a 4 dígitos): {re.findall(r'\d{2,4}', texto)}")

# {n,} - n ou mais repetições
print(f"Padrão r'\\d{{3,}}' (3 ou mais dígitos): {re.findall(r'\d{3,}', texto)}")

print("\n" + "=" * 60)
print("3. CLASSES DE CARACTERES")
print("=" * 60)

texto = "Meu telefone: (11) 98765-4321, CPF: 123.456.789-00"
print(f"\nTexto: {texto}")

# \d - qualquer dígito [0-9]
print(f"\\d (dígitos): {re.findall(r'\d', texto)}")

# \D - qualquer não-dígito
print(f"\\D (não-dígitos): {re.findall(r'\D+', texto)[:5]}")

# \w - caractere alfanumérico [a-zA-Z0-9_]
print(f"\\w+ (palavras): {re.findall(r'\w+', texto)}")

# \W - não-alfanumérico
print(f"\\W (não-alfanuméricos): {re.findall(r'\W+', texto)}")

# \s - espaço em branco (espaço, tab, newline)
print(f"\\s (espaços): quantidade = {len(re.findall(r'\s', texto))}")

# \S - não-espaço em branco
print(f"\\S+ (não-espaços): {re.findall(r'\S+', texto)[:5]}")

print("\n" + "=" * 60)
print("4. CONJUNTOS E RANGES")
print("=" * 60)

texto = "Python3, Java8, C++11, JavaScript2023"
print(f"\nTexto: {texto}")

# [abc] - qualquer um dos caracteres
print(f"Padrão r'[PJC]+': {re.findall(r'[PJC]+', texto)}")

# [a-z] - range de caracteres
print(f"Padrão r'[A-Z][a-z]+': {re.findall(r'[A-Z][a-z]+', texto)}")

# [^abc] - negação - qualquer caractere exceto a, b, c
print(f"Padrão r'[^0-9]+' (sem números): {re.findall(r'[^0-9,]+', texto)}")

print("\n" + "=" * 60)
print("5. GRUPOS E CAPTURA")
print("=" * 60)

texto = "João: 30 anos, Maria: 25 anos, Pedro: 35 anos"
print(f"\nTexto: {texto}")

# () - grupo de captura
padrao = r'(\w+): (\d+) anos'
matches = re.findall(padrao, texto)
print(f"\nPadrão r'(\\w+): (\\d+) anos':")
for nome, idade in matches:
    print(f"  Nome: {nome}, Idade: {idade}")

# (?:...) - grupo não-capturado
texto = "http://site.com https://outro.com.br"
padrao = r'(?:http|https)://(\S+)'
print(f"\nTexto: {texto}")
print(f"Padrão r'(?:http|https)://(\\S+)': {re.findall(padrao, texto)}")

# Backreference - referência a grupos anteriores
texto = "palavra palavra, teste teste, abc abc"
padrao = r'\b(\w+) \1\b'  # \1 referencia o primeiro grupo
print(f"\nTexto: {texto}")
print(f"Padrão r'\\b(\\w+) \\1\\b' (palavras repetidas): {re.findall(padrao, texto)}")

print("\n" + "=" * 60)
print("6. LOOKAHEAD E LOOKBEHIND")
print("=" * 60)

# Lookahead positivo (?=...)
texto = "Python3 Java8 Ruby C#"
print(f"\nTexto: {texto}")
print(f"Padrão r'\\w+(?=\\d)' (palavra seguida de número): {re.findall(r'\w+(?=\d)', texto)}")

# Lookahead negativo (?!...)
print(f"Padrão r'\\w+(?!\\d)\\b' (palavra NÃO seguida de número): {re.findall(r'\w+(?!\d)\b', texto)}")

# Lookbehind positivo (?<=...)
texto = "R$ 100, US$ 50, € 75"
print(f"\nTexto: {texto}")
print(f"Padrão r'(?<=\\$ )\\d+' (números após $): {re.findall(r'(?<=\$ )\d+', texto)}")

# Lookbehind negativo (?<!...)
texto = "test123 abc456 xyz"
print(f"\nTexto: {texto}")
print(f"Padrão r'(?<!\\d)\\w+' (palavras NÃO precedidas por dígito): {re.findall(r'(?<!\d)\w+', texto)}")

print("\n" + "=" * 60)
print("7. FLAGS (MODIFICADORES)")
print("=" * 60)

texto = "Python PYTHON python PyThOn"
print(f"\nTexto: {texto}")

# re.IGNORECASE (re.I) - ignora maiúsculas/minúsculas
print(f"Padrão r'python' com re.I: {re.findall(r'python', texto, re.IGNORECASE)}")

# re.MULTILINE (re.M) - ^ e $ funcionam em cada linha
texto_multiline = """Primeira linha
Segunda linha
Terceira linha"""
print(f"\nTexto multilinha:")
print(texto_multiline)
print(f"Padrão r'^\\w+' com re.M: {re.findall(r'^\w+', texto_multiline, re.MULTILINE)}")

# re.DOTALL (re.S) - . corresponde a qualquer caractere, incluindo \n
texto = "linha1\nlinha2"
print(f"\nTexto: {repr(texto)}")
print(f"Padrão r'linha1.linha2' com re.S: {re.findall(r'linha1.linha2', texto, re.DOTALL)}")

# re.VERBOSE (re.X) - permite comentários e espaços no padrão
padrao_verbose = r'''
    \d{3}     # três dígitos
    [-.]      # traço ou ponto
    \d{3}     # três dígitos
    [-.]      # traço ou ponto
    \d{3}     # três dígitos
    [-.]      # traço ou ponto
    \d{2}     # dois dígitos
'''
texto = "CPF: 123.456.789-00 ou 123-456-789-00"
print(f"\nTexto: {texto}")
print(f"CPF com padrão verbose: {re.findall(padrao_verbose, texto, re.VERBOSE)}")

print("\n" + "=" * 60)
print("8. MÉTODOS PRINCIPAIS DO MÓDULO re")
print("=" * 60)

texto = "Python é incrível! Python é poderoso!"
print(f"\nTexto: {texto}")

# re.search() - encontra a primeira ocorrência
resultado = re.search(r'Python', texto)
if resultado:
    print(f"\nre.search('Python'): encontrado na posição {resultado.start()}-{resultado.end()}")
    print(f"  Texto encontrado: '{resultado.group()}'")

# re.match() - verifica se o início da string corresponde
resultado = re.match(r'Python', texto)
print(f"\nre.match('Python'): {'Match!' if resultado else 'Sem match'}")

resultado = re.match(r'Java', texto)
print(f"re.match('Java'): {'Match!' if resultado else 'Sem match'}")

# re.findall() - encontra todas as ocorrências
print(f"\nre.findall('Python'): {re.findall(r'Python', texto)}")

# re.finditer() - iterador de matches
print(f"\nre.finditer('Python'):")
for match in re.finditer(r'Python', texto):
    print(f"  Posição {match.start()}-{match.end()}: '{match.group()}'")

# re.sub() - substituição
novo_texto = re.sub(r'Python', 'Java', texto)
print(f"\nre.sub('Python', 'Java'): {novo_texto}")

# re.sub() com função
def upper_case(match):
    return match.group().upper()

novo_texto = re.sub(r'\b\w+\b', upper_case, "python é legal", count=2)
print(f"\nre.sub com função (2 primeiras palavras): {novo_texto}")

# re.split() - dividir string
texto = "um,dois;tres:quatro|cinco"
print(f"\nTexto: {texto}")
partes = re.split(r'[,;:|]', texto)
print(f"re.split(r'[,;:|]'): {partes}")

# re.compile() - compilar padrão para reutilização
padrao_compilado = re.compile(r'\d+')
texto = "Tenho 2 gatos e 3 cachorros"
print(f"\nTexto: {texto}")
print(f"Padrão compilado \\d+: {padrao_compilado.findall(texto)}")

print("\n" + "=" * 60)
print("9. GRUPOS NOMEADOS")
print("=" * 60)

texto = "João Silva - joao@email.com, Maria Santos - maria@email.com"
print(f"\nTexto: {texto}")

# (?P<nome>...) - grupo nomeado
padrao = r'(?P<nome>\w+ \w+) - (?P<email>\S+@\S+)'
for match in re.finditer(padrao, texto):
    print(f"  Nome: {match.group('nome')}, Email: {match.group('email')}")

print("\n" + "=" * 60)
print("10. EXEMPLOS PRÁTICOS COMPLEXOS")
print("=" * 60)

# Validar e extrair email
texto = "Contatos: joao@empresa.com.br, maria123@site.co.uk, admin@test"
print(f"\nTexto: {texto}")
padrao_email = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails_validos = re.findall(padrao_email, texto)
print(f"Emails válidos: {emails_validos}")

# Validar senha forte
senhas = ["Senha123!", "senha", "SENHA123", "Senh@123", "S3nh@Forte!"]
padrao_senha = r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$'
print(f"\nValidação de senhas (min 8 chars, maiúsc, minúsc, número, especial):")
for senha in senhas:
    valida = "✓" if re.match(padrao_senha, senha) else "✗"
    print(f"  {valida} {senha}")

# Extrair URLs
texto = "Visite http://site.com ou https://www.exemplo.com.br/pagina?param=valor"
print(f"\nTexto: {texto}")
padrao_url = r'https?://(?:www\.)?[\w\-]+\.[\w\-.]+(?:/[\w\-./?%&=]*)?'
urls = re.findall(padrao_url, texto)
print(f"URLs encontradas: {urls}")

# Extrair hashtags
texto = "Adorei #Python #MachineLearning e #AI! #regex é incrível"
print(f"\nTexto: {texto}")
hashtags = re.findall(r'#\w+', texto)
print(f"Hashtags: {hashtags}")

# Formatar telefone brasileiro
telefones = ["11987654321", "21912345678", "85987654321"]
print(f"\nTelefones: {telefones}")
for tel in telefones:
    tel_formatado = re.sub(r'(\d{2})(\d{5})(\d{4})', r'(\1) \2-\3', tel)
    print(f"  {tel} -> {tel_formatado}")

# Extrair dados de data
texto = "Eventos: 25/12/2023, 01-01-2024, 2024/06/15"
print(f"\nTexto: {texto}")
padrao_data = r'\b(\d{2})[/-](\d{2})[/-](\d{4})\b|\b(\d{4})[/-](\d{2})[/-](\d{2})\b'
datas = re.findall(padrao_data, texto)
print(f"Datas encontradas: {[d for d in datas if any(d)]}")

# Remover tags HTML
html = "<p>Texto <b>importante</b> em <i>HTML</i></p>"
print(f"\nHTML: {html}")
texto_limpo = re.sub(r'<[^>]+>', '', html)
print(f"Texto limpo: {texto_limpo}")

# Validar CPF (formato)
cpfs = ["123.456.789-00", "123456789-00", "123.456.78900", "111.111.111-11"]
print(f"\nValidação de formato de CPF:")
padrao_cpf = r'^\d{3}\.\d{3}\.\d{3}-\d{2}$'
for cpf in cpfs:
    valido = "✓" if re.match(padrao_cpf, cpf) else "✗"
    print(f"  {valido} {cpf}")

# Extrair valores monetários
texto = "Preços: R$ 1.500,00, US$ 200.50, € 50,75, R$300"
print(f"\nTexto: {texto}")
valores = re.findall(r'[R$€US$]+\s*\d+[.,]?\d*[.,]?\d*', texto)
print(f"Valores monetários: {valores}")

print("\n" + "=" * 60)
print("11. GREEDY vs NON-GREEDY (GANANCIOSO vs NÃO-GANANCIOSO)")
print("=" * 60)

texto = "<div>conteúdo1</div><div>conteúdo2</div>"
print(f"\nTexto: {texto}")

# Greedy (padrão) - pega o máximo possível
print(f"Greedy r'<div>.*</div>': {re.findall(r'<div>.*</div>', texto)}")

# Non-greedy - pega o mínimo possível (adicionar ?)
print(f"Non-greedy r'<div>.*?</div>': {re.findall(r'<div>.*?</div>', texto)}")

print("\n" + "=" * 60)
print("12. ESCAPANDO CARACTERES ESPECIAIS")
print("=" * 60)

texto = "O preço é $100. Desconto de 10%! (veja aqui)"
print(f"\nTexto: {texto}")

# Usar \ para escapar caracteres especiais
print(f"Padrão r'\\$\\d+': {re.findall(r'\$\d+', texto)}")
print(f"Padrão r'\\d+%': {re.findall(r'\d+%', texto)}")
print(f"Padrão r'\\(.*?\\)': {re.findall(r'\(.*?\)', texto)}")

# re.escape() - escapa todos os caracteres especiais
texto_especial = "1+1=2, 2*2=4"
print(f"\nTexto: {texto_especial}")
busca = "1+1"
print(f"Buscar '{busca}' sem escape: {re.findall(busca, texto_especial)}")
print(f"Buscar '{busca}' com re.escape(): {re.findall(re.escape(busca), texto_especial)}")

print("\n" + "=" * 60)
print("✅ GUIA COMPLETO DE REGEX CONCLUÍDO!")
print("=" * 60)
```
