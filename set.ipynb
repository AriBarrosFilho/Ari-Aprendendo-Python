## 5. Métodos de Comparação e Testes

```python
# 5.2 issubset() ou <= - Verifica se é subconjunto
A = {1, 2}
B = {1, 2, 3, 4}
C = {5, 6}

print("A:", A)
print("B:", B)
print("C:", C)

print("\nA é subconjunto de B?", A.issubset(B))
print("A <= B:", A <= B)
print("B é subconjunto de A?", B.issubset(A))
print("C é subconjunto de B?", C.issubset(B))

# Subconjunto próprio (< - estritamente menor)
print("\nA é subconjunto próprio de B?", A < B)
print("A é subconjunto próprio de A?", A < A)  # False, pois são iguais
```

```python
# 5.3 issuperset() ou >= - Verifica se é superconjunto
A = {1, 2, 3, 4}
B = {2, 3}
C = {5, 6}

print("A:", A)
print("B:", B)
print("C:", C)

print("\nA é superconjunto de B?", A.issuperset(B))
print("A >= B:", A >= B)
print("B é superconjunto de A?", B.issuperset(A))
print("A é superconjunto de C?", A.issuperset(C))

# Superconjunto próprio (> - estritamente maior)
print("\nA é superconjunto próprio de B?", A > B)
print("A é superconjunto próprio de A?", A > A)  # False, pois são iguais
```

```python
# 5.4 isdisjoint() - Verifica se não há elementos em comum
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 4, 5}

print("A:", A)
print("B:", B)
print("C:", C)

print("\nA e B são disjuntos?", A.isdisjoint(B))
print("A e C são disjuntos?", A.isdisjoint(C))
print("B e C são disjuntos?", B.isdisjoint(C))

# Sets disjuntos têm interseção vazia
print("\nA & B == set():", A & B == set())
```

```python
# 6.2 min(), max(), sum() - Operações com sets numéricos
numeros = {10, 5, 20, 15, 3}

print("Set:", numeros)
print("Mínimo:", min(numeros))
print("Máximo:", max(numeros))
print("Soma:", sum(numeros))
print("Média:", sum(numeros) / len(numeros))

# Funciona também com strings (ordem alfabética)
palavras = {'zebra', 'abelha', 'macaco'}
print("\nSet de palavras:", palavras)
print("Primeira alfabeticamente:", min(palavras))
print("Última alfabeticamente:", max(palavras))
```

## 7. Iteração sobre Sets

## 8. Set Comprehension

```python
# 8.1 Set comprehension básico
quadrados = {x**2 for x in range(10)}
print("Quadrados de 0 a 9:", quadrados)

# Com condição
pares = {x for x in range(20) if x % 2 == 0}
print("\nNúmeros pares de 0 a 19:", pares)

# Note que duplicatas são automaticamente removidas
duplicatas = {x % 5 for x in range(20)}
print("\nResto da divisão por 5 (0-19):", duplicatas)
```

```python
# 8.2 Set comprehension com strings
frase = "Python é uma linguagem de programação"
vogais = {letra for letra in frase.lower() if letra in 'aeiou'}
print("Vogais únicas na frase:", vogais)

# Primeira letra de cada palavra
palavras = "o rato roeu a roupa do rei de roma"
primeiras_letras = {palavra[0] for palavra in palavras.split()}
print("\nPrimeiras letras únicas:", primeiras_letras)
```

## 9. Frozenset (Set Imutável)

```python
# 9.1 Criando frozensets
fs = frozenset([1, 2, 3, 4, 5])
print("Frozenset:", fs)
print("Tipo:", type(fs))

# Frozensets não podem ser modificados
# fs.add(6)  # AttributeError: 'frozenset' object has no attribute 'add'
# fs.remove(1)  # AttributeError

# Mas operações que retornam novos sets funcionam
fs2 = frozenset([4, 5, 6, 7])
print("\nUnião:", fs | fs2)
print("Interseção:", fs & fs2)
```

```python
# 9.2 Frozensets como elementos de sets (sets são mutáveis e não podem ser elementos)
# set_errado = {{1, 2}, {3, 4}}  # TypeError: unhashable type: 'set'

# Mas frozensets podem ser elementos de sets!
set_de_frozensets = {
    frozenset([1, 2]),
    frozenset([3, 4]),
    frozenset([5, 6])
}
print("Set de frozensets:", set_de_frozensets)

# Frozensets também podem ser chaves de dicionários
dicionario = {
    frozenset([1, 2]): "conjunto A",
    frozenset([3, 4]): "conjunto B"
}
print("\nDicionário com frozensets como chaves:", dicionario)
```

## 10. Casos de Uso Práticos

```python
# 10.1 Remover duplicatas de uma lista
lista_com_duplicatas = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5]
print("Lista original:", lista_com_duplicatas)

# Método 1: Converter para set e voltar para lista
lista_sem_duplicatas = list(set(lista_com_duplicatas))
print("Sem duplicatas (pode perder ordem):", lista_sem_duplicatas)

# Método 2: Preservar ordem (Python 3.7+)
def remover_duplicatas_preservando_ordem(lista):
    visto = set()
    resultado = []
    for item in lista:
        if item not in visto:
            visto.add(item)
            resultado.append(item)
    return resultado

lista_sem_dup_ordenada = remover_duplicatas_preservando_ordem(lista_com_duplicatas)
print("Sem duplicatas (ordem preservada):", lista_sem_dup_ordenada)
```

```python
# 10.2 Encontrar elementos comuns entre listas
lista1 = [1, 2, 3, 4, 5]
lista2 = [4, 5, 6, 7, 8]
lista3 = [5, 6, 7, 8, 9]

# Elementos em todas as listas
comuns = set(lista1) & set(lista2) & set(lista3)
print("Elementos comuns a todas:", comuns)

# Elementos em pelo menos duas listas
pelo_menos_duas = (set(lista1) & set(lista2)) | (set(lista1) & set(lista3)) | (set(lista2) & set(lista3))
print("Elementos em pelo menos duas listas:", pelo_menos_duas)
```

```python
# 10.3 Verificar se há elementos únicos em cada lista
vendas_janeiro = {'produto_a', 'produto_b', 'produto_c'}
vendas_fevereiro = {'produto_b', 'produto_c', 'produto_d'}
vendas_marco = {'produto_c', 'produto_d', 'produto_e'}

print("Vendas janeiro:", vendas_janeiro)
print("Vendas fevereiro:", vendas_fevereiro)
print("Vendas março:", vendas_marco)

# Produtos vendidos apenas em janeiro
apenas_janeiro = vendas_janeiro - vendas_fevereiro - vendas_marco
print("\nApenas em janeiro:", apenas_janeiro)

# Produtos vendidos em todos os meses
todos_meses = vendas_janeiro & vendas_fevereiro & vendas_marco
print("Em todos os meses:", todos_meses)

# Produtos vendidos em qualquer mês
qualquer_mes = vendas_janeiro | vendas_fevereiro | vendas_marco
print("Em qualquer mês:", qualquer_mes)
```

```python
# 10.4 Contar palavras únicas em um texto
texto = """
Python é uma linguagem de programação poderosa e versátil.
Python é usada em ciência de dados, desenvolvimento web e automação.
A comunidade Python é muito acolhedora e colaborativa.
"""

palavras = texto.lower().split()
palavras_unicas = set(palavras)

print(f"Total de palavras: {len(palavras)}")
print(f"Palavras únicas: {len(palavras_unicas)}")
print(f"Palavras únicas: {sorted(palavras_unicas)}")
```

```python
# 10.5 Verificar permissões de usuário
usuarios_admin = {'alice', 'bob', 'charlie'}
usuarios_editor = {'bob', 'charlie', 'diana', 'eve'}
usuarios_viewer = {'eve', 'frank', 'grace'}

# Usuário com múltiplas permissões
usuario = 'bob'
permissoes = []
if usuario in usuarios_admin:
    permissoes.append('admin')
if usuario in usuarios_editor:
    permissoes.append('editor')
if usuario in usuarios_viewer:
    permissoes.append('viewer')

print(f"Permissões de {usuario}: {permissoes}")

# Usuários com pelo menos uma permissão
todos_usuarios = usuarios_admin | usuarios_editor | usuarios_viewer
print(f"\nTodos os usuários: {todos_usuarios}")

# Usuários apenas com permissão de visualização
apenas_viewers = usuarios_viewer - usuarios_admin - usuarios_editor
print(f"Apenas viewers: {apenas_viewers}")
```

## 11. Limitações e Considerações Importantes

```python
# 11.1 Sets só aceitam elementos hashable (imutáveis)
# Funcionam: números, strings, tuplas
set_valido = {1, 'texto', (1, 2, 3)}
print("Set válido:", set_valido)

# Não funcionam: listas, sets, dicionários
try:
    set_invalido = {1, 2, [3, 4]}  # TypeError
except TypeError as e:
    print(f"\nErro com lista: {e}")

try:
    set_invalido = {1, 2, {3, 4}}  # TypeError
except TypeError as e:
    print(f"Erro com set: {e}")

# Solução: usar tuplas ao invés de listas, ou frozensets ao invés de sets
set_com_tuplas = {1, 2, (3, 4)}
print("\nSet com tuplas:", set_com_tuplas)

set_com_frozensets = {frozenset([1, 2]), frozenset([3, 4])}
print("Set com frozensets:", set_com_frozensets)
```

```python
# 11.2 Sets não mantêm ordem
# A partir do Python 3.7, dicionários mantêm ordem de inserção, mas sets NÃO!
numeros = {5, 1, 3, 2, 4}
print("Set criado com {5, 1, 3, 2, 4}:", numeros)
print("Ordem não é garantida!")

# Se precisar de ordem, use lista ou considere outras estruturas
from collections import OrderedDict
# Ou simplesmente use sorted() quando iterar
```

```python
# 11.3 Comparação de eficiência: Set vs Lista
import time

# Criando estruturas grandes
tamanho = 10000
lista = list(range(tamanho))
conjunto = set(range(tamanho))

# Teste de pertencimento
elemento_buscar = tamanho - 1

# Lista
inicio = time.time()
for _ in range(1000):
    elemento_buscar in lista
tempo_lista = time.time() - inicio

# Set
inicio = time.time()
for _ in range(1000):
    elemento_buscar in conjunto
tempo_set = time.time() - inicio

print(f"Tempo com lista: {tempo_lista:.4f}s")
print(f"Tempo com set: {tempo_set:.4f}s")
print(f"Set é {tempo_lista/tempo_set:.1f}x mais rápido para busca!")

print("\n⚠️ Use sets quando:")
print("  • Precisa de unicidade")
print("  • Faz muitas buscas de pertencimento")
print("  • Precisa de operações de conjuntos")
print("\n⚠️ Use listas quando:")
print("  • Precisa manter ordem")
print("  • Permite duplicatas")
print("  • Precisa acessar por índice")
```

## 12. Resumo de Métodos e Operadores

| Operação | Método | Operador | Descrição |
|----------|--------|----------|-----------|
| União | `union()` | `\|` | Todos os elementos |
| Interseção | `intersection()` | `&` | Elementos comuns |
| Diferença | `difference()` | `-` | Elementos em A mas não em B |
| Diferença Simétrica | `symmetric_difference()` | `^` | Elementos em A ou B, não ambos |
| União in-place | `update()` | `\|=` | Modifica o set original |
| Interseção in-place | `intersection_update()` | `&=` | Modifica o set original |
| Diferença in-place | `difference_update()` | `-=` | Modifica o set original |
| Dif. Simétrica in-place | `symmetric_difference_update()` | `^=` | Modifica o set original |
| Subconjunto | `issubset()` | `<=` | A está contido em B |
| Subconjunto próprio | - | `<` | A está contido em B (diferente) |
| Superconjunto | `issuperset()` | `>=` | A contém B |
| Superconjunto próprio | - | `>` | A contém B (diferente) |
| Disjunto | `isdisjoint()` | - | Sem elementos comuns |

```python
# Exercício 1: Análise de dados de e-commerce
# Três clientes fizeram compras em diferentes dias
compras_cliente1 = {'notebook', 'mouse', 'teclado', 'webcam'}
compras_cliente2 = {'mouse', 'mousepad', 'headset', 'webcam'}
compras_cliente3 = {'teclado', 'mouse', 'monitor'}

# 1. Quais produtos foram comprados por todos os clientes?
todos = compras_cliente1 & compras_cliente2 & compras_cliente3
print("Produtos comprados por todos:", todos)

# 2. Quais produtos únicos cada cliente comprou?
unicos_c1 = compras_cliente1 - compras_cliente2 - compras_cliente3
unicos_c2 = compras_cliente2 - compras_cliente1 - compras_cliente3
unicos_c3 = compras_cliente3 - compras_cliente1 - compras_cliente2
print(f"Únicos cliente 1: {unicos_c1}")
print(f"Únicos cliente 2: {unicos_c2}")
print(f"Únicos cliente 3: {unicos_c3}")

# 3. Catálogo completo de produtos vendidos
catalogo = compras_cliente1 | compras_cliente2 | compras_cliente3
print(f"\nCatálogo completo ({len(catalogo)} produtos): {catalogo}")
```
